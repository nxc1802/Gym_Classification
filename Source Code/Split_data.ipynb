```python
# Iterate through the subdirectories (class folders)
for class_folder in os.listdir(root_dir):
    class_path = os.path.join(root_dir, class_folder)

    # Check if it's a directory
    if os.path.isdir(class_path):
        print(f"Class: {class_folder}")
        video_files = []
        # Iterate through files in the class folder
        for filename in os.listdir(class_path):
            file_path = os.path.join(class_path, filename)
            if os.path.isfile(file_path) and filename.lower().endswith(('.mp4', '.mov', '.avi')):
                video_files.append(filename)

        num_videos = len(video_files)
        if num_videos == 0:
            print("No video files found in this class folder.")
            continue

        num_train = math.floor(num_videos * 0.5)
        num_val = math.ceil(num_videos * 0.2)
        num_test = math.ceil(num_videos * 0.3)


        random.shuffle(video_files) # Shuffle before splitting


        train_files = video_files[:num_train]
        val_files = video_files[num_train:num_train + num_val]
        test_files = video_files[num_train + num_val:]

        print(f"  - Train: {len(train_files)} files")
        print(f"  - Val: {len(val_files)} files")
        print(f"  - Test: {len(test_files)} files")

        # Example: Print the filenames for each split
        print("Train Files:", train_files)
        print("Validation Files:", val_files)
        print("Test Files:", test_files)
```

```python
# Specify the root directory and the destination directory
root_dir = '/content/drive/MyDrive/Gym/data/Merge_data'
dest_dir = '/content/drive/MyDrive/Gym/data/Marge_data_split'

# Create destination directory if it doesn't exist
os.makedirs(dest_dir, exist_ok=True)

# Iterate through the subdirectories (class folders)
for class_folder in os.listdir(root_dir):
    class_path = os.path.join(root_dir, class_folder)

    if os.path.isdir(class_path):
        print(f"Processing class: {class_folder}")
        video_files = []
        for filename in os.listdir(class_path):
            file_path = os.path.join(class_path, filename)
            if os.path.isfile(file_path) and filename.lower().endswith(('.mp4', '.mov', '.avi')):
                video_files.append(filename)

        num_videos = len(video_files)
        if num_videos == 0:
            print("No video files found in this class folder.")
            continue

        num_train = math.floor(num_videos * 0.6)
        num_val = math.ceil(num_videos * 0.2)
        num_test = num_videos - num_train - num_val


        random.shuffle(video_files)

        train_files = video_files[:num_train]
        val_files = video_files[num_train:num_train + num_val]
        test_files = video_files[num_train + num_val:]

        for split, files in zip(['train', 'val', 'test'], [train_files, val_files, test_files]):
            split_dir = os.path.join(dest_dir, split, class_folder)
            os.makedirs(split_dir, exist_ok=True)
            for file in files:
              source_path = os.path.join(class_path,file)
              destination_path = os.path.join(split_dir,file)
              shutil.copy2(source_path,destination_path)

        print(f"  - Train: {len(train_files)} files")
        print(f"  - Val: {len(val_files)} files")
        print(f"  - Test: {len(test_files)} files")
```

```python
# prompt: Thực hiện split tương tự như trên nhưng là với các csv chứ k phải mp4

from google.colab import drive
import os
import math
import random
import shutil

drive.mount('/content/drive')

# Specify the root directory and the destination directory
root_dir = '/content/drive/MyDrive/Gym/data/Merge_data_to_csv_unnormalize'
dest_dir = '/content/drive/MyDrive/Gym/data/Merge_data_to_csv_unnormalize_split'

# Create destination directory if it doesn't exist
os.makedirs(dest_dir, exist_ok=True)

# Iterate through the subdirectories (class folders)
for class_folder in os.listdir(root_dir):
    class_path = os.path.join(root_dir, class_folder)

    if os.path.isdir(class_path):
        print(f"Processing class: {class_folder}")
        csv_files = []  # Changed to csv_files
        for filename in os.listdir(class_path):
            file_path = os.path.join(class_path, filename)
            if os.path.isfile(file_path) and filename.lower().endswith('.csv'):  # Check for CSV files
                csv_files.append(filename)

        num_files = len(csv_files)  # Changed to num_files
        if num_files == 0:
            print("No CSV files found in this class folder.")
            continue

        num_train = math.floor(num_files * 0.6)
        num_val = math.ceil(num_files * 0.2)
        num_test = num_files - num_train - num_val

        random.shuffle(csv_files)  # Shuffle the CSV files

        train_files = csv_files[:num_train]
        val_files = csv_files[num_train:num_train + num_val]
        test_files = csv_files[num_train + num_val:]

        for split, files in zip(['train', 'val', 'test'], [train_files, val_files, test_files]):
            split_dir = os.path.join(dest_dir, split, class_folder)
            os.makedirs(split_dir, exist_ok=True)
            for file in files:
                source_path = os.path.join(class_path, file)
                destination_path = os.path.join(split_dir, file)
                shutil.copy2(source_path, destination_path)

        print(f"  - Train: {len(train_files)} files")
        print(f"  - Val: {len(val_files)} files")
        print(f"  - Test: {len(test_files)} files")
```
